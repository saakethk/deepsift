# Exploring LLM Prompting

```python
# %pip install langchain_ollama
from langchain_ollama import ChatOllama
```

```python
# Initialize Gemma
llm = ChatOllama(
    model="gemma3:12b",
    temperature=0.2,
)
```

```python
# Test a simple prompt
for chunk in llm.stream("Explain what an ATS-friendly resume is in 2 sentences:"):
    print(chunk.content, end="", flush=True)
```
